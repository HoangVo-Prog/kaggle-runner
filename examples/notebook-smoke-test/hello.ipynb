# Kaggle Runner notebook example
This notebook writes a small JSON artifact under KGR_OUTPUT_DIR.

```python
import json, os, pathlib, platform
output = pathlib.Path(os.environ.get('KGR_OUTPUT_DIR', '.'))
output.mkdir(parents=True, exist_ok=True)
payload = {'status': 'ok', 'python': platform.python_version()}
(output / 'notebook_result.json').write_text(json.dumps(payload, indent=2))
print(payload)
```
